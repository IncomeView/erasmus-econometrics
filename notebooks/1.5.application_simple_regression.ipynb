# 2. O Modelo de Vendas e Preço

O modelo utilizado é:

$$

O coeficiente:

$$

$$

$$

Em muitas aplicações o intercepto possui

# 5. Interpretando o Coeficiente Angular

O coeficiente:

$$

$$

$$

$$

$$

$$

# 22. Modelo Linear

# 27. Limitação da Tendência Linear

O modelo linear assume:

$$
W_i
=
\alpha+\beta G_i+\varepsilon_i
$$

Isso implica que a redução do tempo vencedor
é constante ao longo do tempo.

Se extrapolarmos indefinidamente:

$$
W_i
\rightarrow 0
$$

para valores suficientemente grandes de:

$$
G
$$

O resultado é claramente implausível.

Nenhum ser humano pode correr 100 metros em zero segundos.

Esse é o principal argumento utilizado pelo professor
para apresentar um modelo alternativo.

【1-9ed8b2】

# 28. Modelo com Ganhos Relativos Constantes

A aula propõe:

$$
W_i
=
\gamma
e^{\beta G_i}
$$

Esse modelo implica reduções percentuais constantes
ao longo do tempo.

Aplicando logaritmo:

$$
\log(W_i)
=
\alpha
+
\beta G_i
+
\varepsilon_i
$$

onde:

$$
\alpha=\log(\gamma)
$$

【1-9ed8b2】

# Questão (a)

Mostrar que o modelo linear prevê tempos iguais
aproximadamente no ano de 2140.

Vamos utilizar os modelos lineares estimados anteriormente.

## Modelo Linear para Homens

Da regressão:

$$
W_M
=
a_M
+
b_MG
$$

onde:

$$
a_M
=
\text{intercepto do modelo masculino}
$$

$$
b_M
=
\text{inclinação do modelo masculino}
$$

```python
a_m = model_men.params["const"]
b_m = model_men.params["Game"]

a_m, b_m
```

## Modelo Linear para Mulheres

Da regressão:

$$
W_W
=
a_W
+
b_WG
$$

```python
a_w = model_women.params["const"]
b_w = model_women.params["Game"]

a_w, b_w
```

## Determinando o Ponto de Interseção

Tempos iguais significam:

$$
W_M=W_W
$$

Portanto:

$$
a_M+b_MG
=
a_W+b_WG
$$

Rearranjando:

$$
G
=
\frac{
a_W-a_M
}{
b_M-b_W
}
$$

```python
G_equal_linear = (
    (a_w - a_m)
    /
    (b_m - b_w)
)

G_equal_linear
```

O resultado está em unidades de Olimpíadas.

Para converter em ano calendário:

$$
Year
=
1948
+
4(G-1)
$$

```python
year_equal_linear = (
    1948
    +
    4*(G_equal_linear - 1)
)

year_equal_linear
```

Compare o resultado obtido
com a afirmação apresentada no exercício.

\log(W_W)
=

B_WG
$$

```python
A_w = model_log_women.params["const"]
B_w = model_log_women.params["Game"]

A_w, B_w
```

## Igualando os Tempos

Tempos iguais implicam:

$$
W_M=W_W
$$

Como a função logaritmo é monotônica:

$$
\log(W_M)

A_M+B_MG

G

\frac{
A_W-A_M
}{
B_M-B_W

```python
G_equal_nonlinear = (
    (A_w - A_m)
    /
    (B_m - B_w)
)

G_equal_nonlinear
```

```python
year_equal_nonlinear = (
    1948
    +
    4*(G_equal_nonlinear - 1)
)

year_equal_nonlinear
```
