# 1. A Linha de Regressão

Considere:

$$
n
$$

observações formadas por pares:

$$
(x_i,y_i)
$$

Nosso objetivo é ajustar uma reta:

$$
y=a+bx
$$

aos dados observados.

Essa reta é chamada: **Linha de Regressão**


# 2. Valores Ajustados

A reta fornece para cada observação:

$$
x_i
$$

um valor previsto:

$$
\hat y_i
=
a+bx_i
$$

Esses valores representam a parte explicada
da variável:

$$
y
$$

pela variável:

$$
x
$$



# 3. Resíduos

Nem todos os pontos estarão exatamente sobre a reta.

Definimos:

$$
e_i
=
y_i-\hat y_i
$$

ou:

$$
e_i
=
y_i-a-bx_i
$$

O resíduo mede:

- erro da previsão;
- distância vertical do ponto até a reta.



# 4. O Problema da Estimação

Existem infinitas retas possíveis.

Precisamos de um critério para decidir
qual delas é a melhor.

Intuitivamente:

queremos uma reta que produza resíduos pequenos.

Quanto menores forem os resíduos,
melhor será o ajuste da reta aos dados.


# 5. Método dos Mínimos Quadrados

O critério adotado é:

minimizar a soma dos resíduos ao quadrado.

Definimos:

$$
S(a,b)
=
\sum_{i=1}^{n}
e_i^2
$$

ou:

$$
S(a,b)
=
\sum_{i=1}^{n}
(y_i-a-bx_i)^2
$$

Os valores escolhidos para:

$$
a
$$

e:

$$
b
$$

serão aqueles que minimizam:

$$
S(a,b)
$$


# 6. Intuição do Critério

Por que elevar os resíduos ao quadrado?

Porque:

- elimina sinais negativos;
- penaliza erros grandes;
- gera uma função matemática conveniente.

Assim,

resíduos maiores recebem peso proporcionalmente maior
dentro da função objetivo.



# 7. Condições de Primeira Ordem

O mínimo de:

$$
S(a,b)
$$

é obtido resolvendo:

$$
\frac{\partial S}{\partial a}=0
$$

e:

$$
\frac{\partial S}{\partial b}=0
$$

A lecture começa derivando a solução para:

$$
a
$$



# 8. Derivando a Equação para a

A derivada em relação a:

$$
a
$$

resulta em:

$$
0
=
-2
\sum_{i=1}^{n}
(y_i-a-bx_i)
$$

Mas:

$$
e_i
=
y_i-a-bx_i
$$

Logo:

$$
\boxed{
\sum e_i=0
}
$$

Essa é a primeira propriedade fundamental dos resíduos.



# 9. Interpretação

A soma residual é sempre zero.

Isso significa que:

- resíduos positivos compensam resíduos negativos;
- o modelo não apresenta viés sistemático nos dados ajustados.

Consequentemente:

a média dos resíduos também será igual a zero.


# 10. Obtendo a Fórmula de a

Partimos de:

$$
\sum y_i
-
na
-
b\sum x_i
=
0
$$

Dividindo todos os termos por:

$$
n
$$

obtemos:

$$
\bar y
-
a
-
b\bar x
=
0
$$

Finalmente:

$$
\boxed{
a
=
\bar y-b\bar x
}
$$



# 11. Interpretação de a

O intercepto não é calculado
de forma independente.

Ele depende:

- da média de y;
- da média de x;
- do valor de b.

Isso implica uma propriedade extremamente importante.



# 12. A Reta Passa Pelas Médias

Substituindo:

$$
x=\bar x
$$

em:

$$
\hat y=a+b\bar x
$$

e utilizando:

$$
a=\bar y-b\bar x
$$

obtemos:

$$
\hat y
=
\bar y
$$

Portanto:

$$
\boxed{
(\bar x,\bar y)
}
$$

pertence à reta estimada.


# 13. Questão Conceitual da Aula

Suponha que os dados tenham sido centralizados:

$$
x_i^*
=
x_i-\bar x
$$

$$
y_i^*
=
y_i-\bar y
$$

Nesse caso:

$$
\bar x^*=0
$$

e:

$$
\bar y^*=0
$$

# 14. Resolvendo a Questão

Como:

$$
a
=
\bar y
-

$$

então:

$$
a
=
0
-
b(0)
$$

Logo:

$$
\boxed{

}
$$

independentemente do valor de:

$$

$$

A lecture destaca que:

$$

$$

não é alterado pela centralização dos dados.

# 16. Derivando o Coeficiente Angular b

Agora passamos ao segundo parâmetro da regressão:

$$
b
$$

O objetivo é encontrar o valor de:

$$
b
$$

que minimiza:

$$
S(a,b)
=
\sum
(y_i-a-bx_i)^2
$$

Para isso calculamos:

$$
\frac{\partial S}{\partial b}
=
0
$$


# 17. Primeira Ordem para b

Derivando:

$$
S(a,b)
=
\sum
(y_i-a-bx_i)^2
$$

em relação a:

$$
b
$$

obtemos:

$$
0
=
-2
\sum
x_i
(y_i-a-bx_i)
$$

Como:

$$
e_i
=
y_i-a-bx_i
$$

isso equivale a:

$$
\boxed{
\sum x_ie_i = 0
}
$$

Esta é a segunda propriedade fundamental dos resíduos.


# 18. Interpretando a Propriedade

O resultado:

$$
\sum x_ie_i = 0
$$

significa que:

os resíduos não possuem relação linear
com a variável explicativa.

Dizemos que:

$$
x
$$

é ortogonal aos resíduos.

Essa propriedade será utilizada diversas vezes
ao longo do curso.


# 19. Desenvolvendo a Equação

Partimos de:

$$
\sum x_i(y_i-a-bx_i)
=
0
$$

Expandindo:

$$
\sum x_iy_i
-
a\sum x_i
-
b\sum x_i^2
=
0
$$

Agora utilizamos:

$$
a
=
\bar y
-
b\bar x
$$

obtido anteriormente.


# 20. Substituindo o Valor de a

Temos:

$$
\sum x_iy_i
-
(\bar y-b\bar x)\sum x_i
-
b\sum x_i^2
=
0
$$

Rearranjando os termos:

$$
\sum x_i(y_i-\bar y)
=
b
\sum x_i(x_i-\bar x)
$$

A expressão já possui uma forma muito mais conveniente.


# 21. Obtendo a Primeira Fórmula para b

Portanto:

$$
b
=
\frac{
\sum x_i(y_i-\bar y)
}{
\sum x_i(x_i-\bar x)
}
$$

Esta é uma expressão correta,
mas ainda não é a forma mais utilizada.


# 22. Utilizando uma Propriedade das Médias

Observe:

$$
\sum (y_i-\bar y)
=
0
$$

e:

$$
\sum (x_i-\bar x)
=
0
$$

Portanto:

$$
\bar x
\sum(y_i-\bar y)
=
0
$$

e:

$$
\bar x
\sum(x_i-\bar x)
=
0
$$


# 23. Forma Final de b

Utilizando essas propriedades,
podemos escrever:

$$
\boxed{
b
=
\frac{
\sum
(x_i-\bar x)
(y_i-\bar y)
}{
\sum
(x_i-\bar x)^2
}
}
$$

Esta é a forma mais importante
do estimador de inclinação da regressão simples.


# 24. Interpretação da Fórmula

O numerador:

$$
\sum
(x_i-\bar x)
(y_i-\bar y)
$$

mede o grau de associação entre:

$$
x
$$

e:

$$
y
$$

O denominador:

$$
\sum
(x_i-\bar x)^2
$$

mede a variabilidade existente em:

$$
x
$$

Assim:

o coeficiente angular compara
a covariabilidade entre as variáveis
com a variabilidade existente em:

$$
x
$$


# 25. Covariância e Variância

A fórmula anterior pode ser escrita como:

$$
\boxed{
b
=
\frac{
Cov(x,y)
}{
Var(x)
}
}
$$

Esta forma é extremamente útil
para interpretar o comportamento do estimador.

# 32. Decomposição da Variabilidade

Começamos da expressão:

$$
y_i
=
a+bx_i+e_i
$$

Utilizando:

$$
a
=
\bar y-b\bar x
$$

podemos escrever:

$$
y_i
=
\bar y
-
b\bar x
+
bx_i
+
e_i
$$

Reorganizando:

$$
y_i-\bar y
=
b(x_i-\bar x)
+
e_i
$$


# 33. Interpretando a Expressão

O resultado:

$$
y_i-\bar y
=
b(x_i-\bar x)
+
e_i
$$

possui interpretação importante.

O lado esquerdo:

$$
y_i-\bar y
$$

representa o desvio da observação
em relação à média.

O lado direito mostra que esse desvio possui:

- uma parte explicada pela regressão;
- uma parte não explicada pelos resíduos.


# 34. Uma Propriedade Fundamental

Já demonstramos:

$$
\sum e_i = 0
$$

e:

$$
\sum x_ie_i = 0
$$

Logo:

$$
\sum
(x_i-\bar x)e_i
=
0
$$

Isso significa que:

a parte explicada

e

a parte residual

são ortogonais.


# 35. Elevando ao Quadrado

Partimos novamente de:

$$
y_i-\bar y
=
b(x_i-\bar x)+e_i
$$

Elevando ao quadrado:

$$
(y_i-\bar y)^2
=
b^2(x_i-\bar x)^2
+
e_i^2
+
2b(x_i-\bar x)e_i
$$

Somando todas as observações:

$$
\sum
(y_i-\bar y)^2
=
b^2
\sum
(x_i-\bar x)^2
+
\sum e_i^2
$$

porque:

$$
\sum
(x_i-\bar x)e_i
=
0
$$


# 36. Soma Total dos Quadrados

Definimos:

$$
SST
=
\sum
(y_i-\bar y)^2
$$

conhecida como:

## Soma Total dos Quadrados

Ela mede a variabilidade total existente em:

$$
y
$$


# 37. Soma Explicada e Soma Residual

Definimos:

$$
SSE
=
b^2
\sum
(x_i-\bar x)^2
$$

como a parte explicada pela regressão.

E:

$$
SSR
=
\sum e_i^2
$$

como a parte residual.

Logo:

$$
\boxed{
SST
=
SSE
+
SSR
}
$$


# 38. O Coeficiente de Determinação

A partir da decomposição anterior,
definimos:

$$
\boxed{
R^2
=
\frac{
SSE
}{
SST
}
}
$$

Ou seja:

a fração da variabilidade total
que é explicada pela regressão.


# 39. Forma Alternativa de R²

Como:

$$
SST=SSE+SSR
$$

podemos escrever:

$$
\boxed{
R^2
=
1
-
\frac{
SSR
}{
SST
}
}
$$

ou:

$$
\boxed{
R^2
=
1
-
\frac{
\sum e_i^2
}{
\sum(y_i-\bar y)^2
}
}
$$

Esta é a fórmula apresentada na lecture.


# 40. Interpretando o R²

Se:

$$
R^2=0
$$

então:

a regressão não explica nenhuma parcela
da variabilidade observada.

Se:

$$
R^2=1
$$

então:

todos os pontos estão exatamente sobre a reta.

O ajuste é perfeito.


# 41. Intervalo de Valores Possíveis

Em regressão simples:

$$
0
\le
R^2
\le
1
$$

Quanto maior:

$$
R^2
$$

maior a capacidade explicativa do modelo.


# 42. Variância dos Erros

O modelo populacional assume:

$$

\sim

$$

Entretanto:

$$
\sigma^2
$$

é desconhecida.

# 43. Quantos Resíduos São Livres?

Os resíduos:

$$
e_1,e_2,\ldots,e_n
$$

não são totalmente independentes.

Já demonstramos:

$$
\sum e_i=0
$$

e:

$$
\sum x_ie_i=0
$$

Portanto:

duas restrições são impostas aos resíduos.

Logo:

apenas:

$$
n-2
$$

resíduos podem variar livremente.


# 44. Estimador da Variância dos Erros

Como:

$$
\bar e=0
$$

obtemos:

$$
\boxed{
s^2
=
\frac{
\sum e_i^2
}{
n-2
}
}
$$

Esta é a variância residual estimada.


# 45. Desvio Padrão Residual

Frequentemente utilizamos:

$$
s
=
\sqrt{s^2}
$$

que possui a mesma unidade de medida da variável:

$$
y
$$

Quanto menor:

$$
s
$$

melhor tende a ser o ajuste.


$$SST = SSExplained + SSResidual$$
Pela definição, $$R^2=\frac{SSExplained}{SST}$$
Como $$SSExplained=SST-SSResidual$$

# 50. Construindo as Variáveis

```python
x = df["Game"]
y = df["Winning time men"]
```

# 51. Médias Amostrais

```python
x_bar = x.mean()
y_bar = y.mean()

x_bar, y_bar
```

Essas médias serão utilizadas
diretamente nas fórmulas derivadas na lecture.

# 52. Calculando b

Utilizamos:

$$
b
=
\frac{
\sum
(x_i-\bar x)
(y_i-\bar y)
}{
\sum
(x_i-\bar x)^2
}
$$


```python
numerador = ((x - x_bar) * (y - y_bar)).sum()
denominador = ((x - x_bar)**2).sum()

b = numerador / denominador
b
```

O valor exibido corresponde
ao coeficiente angular estimado.

# 53. Calculando a

Utilizamos:

$$
a
=
\bar y-b\bar x
$$


```python
a = y_bar - b*x_bar
a
```

O valor exibido corresponde
ao intercepto estimado.

# 54. Construindo a Reta Ajustada

```python
y_hat = a + b*x
y_hat.head()
```

Cada observação em:

$$
y_hat
$$

representa o valor previsto pelo modelo.

# 55. Visualizando o Ajuste

```python
plt.figure(figsize=(8,5))

plt.scatter(x, y, label="Observed")

plt.plot(
    x,
    y_hat,
    color="red",
    label="Regression Line"
)

plt.xlabel("Game")
plt.ylabel("Winning Time")

plt.legend()
plt.grid(True)

plt.show()
```

A linha estimada resume
a tendência média observada nos dados.

# 56. Calculando os Resíduos

Lembrando:

$$
e_i
=
y_i-\hat y_i
$$

```python
e = y - y_hat
e.head()
```

# 57. Verificando a Primeira Propriedade dos Resíduos

```python
e.sum()
```

O resultado deve ser extremamente próximo de:

$$
0
$$

conforme demonstrado durante a lecture.


# 58. Verificando a Segunda Propriedade

```python
(x * e).sum()
```

O resultado também deve ser próximo de:

$$
0
$$

$$

$$


# 59. Soma Residual dos Quadrados

```python
SSR = (e**2).sum()
SSR
```

Esta é a quantidade de variabilidade
não explicada pela regressão.

# 60. Soma Total dos Quadrados

```python
SST = ((y - y_bar)**2).sum()
SST
```

Esta é a variabilidade total observada

# 61. Calculando R²

```python
R2 = 1 - SSR/SST
R2
```

Esse valor indica a proporção da variação
dos tempos vencedores
explicada pela tendência linear.


# 62. Calculando s²

```python
n = len(df)
s2 = SSR/(n-2)

s2
```

Utilizamos:

$$

=
\frac{

}{

}

# 63. Calculando s

```python
s = np.sqrt(s2)

s
```
