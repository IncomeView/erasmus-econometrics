## Motivação

Uma das tarefas mais importantes da econometria é prever resultados futuros.

Para isso utilizamos informações observadas no passado.

Em muitos problemas econômicos uma variável pode ajudar a explicar outra.

Exemplos:

- preço e vendas;
- educação e salário;
- investimento e crescimento econômico.

Nesta aula utilizaremos um exemplo simples envolvendo vendas e preço
para introduzir os conceitos fundamentais da regressão.

# 3. Variável Dependente e Variável Explicativa

Em regressão simples trabalhamos com duas variáveis.

## Variável dependente

É a variável que desejamos explicar ou prever.

Neste exemplo:

$$
Y = Sales
$$

---

## Variável explicativa

É a variável utilizada para explicar a variável dependente.

Neste exemplo:

$$
X = Price
$$

---

Também utilizaremos frequentemente os termos:

- variável explicada;
- variável dependente;

e

- variável explicativa;
- variável regressora.

# 4. Uma Relação Linear

Os dados sugerem que uma reta pode representar adequadamente
a relação entre preço e vendas.

Escrevemos:

$$
Sales
=
a+bPrice
$$

onde:

$$
a
$$

representa o intercepto.

E:

$$
b
$$

representa a inclinação da reta.

O coeficiente:

$$
b
$$

mede a variação esperada nas vendas quando o preço aumenta uma unidade.

Em particular:

- $b<0$ → relação negativa;
- $b>0$ → relação positiva;
- $b=0$ → ausência de relação linear.

# 5. Vendas Previstas e Resíduos

A reta produz uma previsão para o nível de vendas.

Denotamos essa previsão por:

$$
\widehat{Sales}
=
a+bPrice
$$

Porém os valores observados raramente coincidem exatamente com a reta.

A diferença entre:

- valor observado;
- valor previsto;

é chamada de resíduo.

$$
e
=
Sales-\widehat{Sales}
$$

Quanto menor o resíduo:

- melhor a previsão.

Quanto maior o resíduo:

- pior a previsão.

# 6. Quando Não Existe Relação

Considere um cenário hipotético em que o preço não influencia as vendas.

Nesse caso:

$$
b=0
$$

e a equação torna-se:

$$
Sales=a
$$

A previsão seria a mesma independentemente do preço.

O preço não forneceria qualquer informação adicional para previsão.

# 7. Previsão Incondicional

Suponha que o preço seja desconhecido.

Nesse caso utilizamos apenas as informações das vendas observadas.

A previsão mais simples é a média amostral:

$$
\bar y
=
\frac1n
\sum_{i=1}^{n}
y_i
$$

Essa previsão é chamada:

## previsão incondicional

porque não utiliza informação adicional.

# 8. Previsão Condicional

Agora suponha que o preço seja conhecido.

Podemos utilizar essa informação para melhorar a previsão.

A previsão passa a depender de:

$$
Price
$$

através da relação:

$$
Sales=a+bPrice
$$

Essa é chamada:

## previsão condicional

porque depende do valor observado de uma segunda variável.

Em geral, previsões condicionais tendem a ser superiores às previsões
baseadas apenas na média.

# 9. Aplicação: Maximização do Faturamento

O faturamento é definido como:

$$
Turnover
=
Price \times Sales
$$

Como:

$$
Sales=a+bPrice
$$

então:

$$
Turnover
=
Price(a+bPrice)
$$

ou:

$$
T(P)
=
aP+bP^2
$$

Para encontrar o preço ótimo:

$$
\frac{dT}{dP}
=
a+2bP
$$

Igualando a zero:

$$
a+2bP=0
$$

obtemos:

$$
P^*
=
-\frac{a}{2b}
$$

Este é o preço que maximiza o faturamento.

# 10. Distribuição Normal

A aula apresenta uma representação simplificada para as vendas:

$$
Y_i
\sim
NID(\mu,\sigma^2)
$$

onde:

- $\mu$ é a média populacional;
- $\sigma^2$ é a variância populacional.

Um estimador da média populacional é:

$$
\bar y
=
\frac1n
\sum_{i=1}^{n}
y_i
$$

Quando nenhuma informação adicional está disponível,
a média constitui a melhor previsão possível.

# 11. Qual Situação É Mais Fácil de Prever?

A aula apresenta três gráficos hipotéticos.

A situação mais favorável para previsão é aquela em que:

- os pontos estão mais próximos da reta;
- os resíduos são menores;
- existe menor variação ao redor do relacionamento médio.

Quanto menor a dispersão ao redor da reta,
maior tende a ser a qualidade da previsão.

## Questão (d)

Calcular a média amostral dos gastos.

## Questão (e)

Calcular:

- média dos clientes com menos de 40 anos;
- média dos clientes com 40 anos ou mais.

```python
young = df[df["Age"] < 40]
old = df[df["Age"] >= 40]

young_mean = young["Expenditures"].mean()
old_mean = old["Expenditures"].mean()

young_mean, old_mean
```

### Resultado

Clientes com menos de 40 anos:

$$
\bar y_{<40}
\approx 106.07
$$

Clientes com 40 anos ou mais:

$$
\bar y_{\ge40}
\approx 96.08
$$

Existe diferença considerável entre os grupos.
