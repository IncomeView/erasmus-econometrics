Considere novamente o exemplo:

A análise visual dos dados sugeriu que:

$$
Price \uparrow

Sales \downarrow

Também vimos que uma reta parecia ajustar razoavelmente bem os pontos

Nesta aula formalizaremos essa ideia.

# 2. Média Incondicional

Começamos pelo caso mais simples.

Suponha que somente observamos:

$$
y_1,y_2,\ldots,y_n
$$

sem nenhuma informação adicional.

Assumimos:

$$
y_i
\sim
NID(\mu,\sigma^2)
$$

onde:

- μ representa a média populacional;
- σ² representa a variância populacional.

Nesse caso:

$$
E(y_i)=\mu
$$

e:

$$
Var(y_i)=\sigma^2
$$

A melhor previsão disponível é simplesmente a média.

# 3. Estimadores Amostrais

Como normalmente não conhecemos os parâmetros populacionais,
precisamos estimá-los utilizando a amostra.

A média populacional é estimada por:

$$
\hat{\mu}
=
\bar y
=
\frac1n
\sum_{i=1}^{n}
y_i
$$

A variância populacional é estimada por:

$$
\hat{\sigma}^2
=
\frac1{n-1}
\sum_{i=1}^{n}
(y_i-\bar y)^2
$$

Esses estimadores resumem as informações presentes na amostra.

# 4. Da Média Incondicional para a Média Condicional

A média incondicional utiliza apenas a variável y.

Entretanto muitas vezes possuímos informação adicional.

Por exemplo:

- preço;
- renda;
- idade;
- educação.

Podemos incorporar essa informação substituindo:

$$
E(y_i)=\mu
$$

por:

$$
E(y_i)
=
\alpha+\beta x_i
$$

Essa expressão afirma que a média esperada de y depende do valor assumido
por x.

Esta é a ideia central da regressão simples.

# 5. O Modelo de Regressão Simples

O modelo de regressão simples é:

$$
E(y_i)
=
\alpha+\beta x_i
$$

onde:

- α é o intercepto;
- β é o coeficiente angular.

O coeficiente β mede o efeito marginal de x sobre y:

$$
\beta
=
\frac{dy}{dx}
$$

Assim:

- β > 0 implica relação positiva;
- β < 0 implica relação negativa;
- β = 0 implica ausência de efeito linear.

# 6. Forma Alternativa do Modelo

O modelo também pode ser escrito na forma:

$$
y_i
=
\alpha
+
\beta x_i
+
\varepsilon_i
$$

onde:

$$
\varepsilon_i
\sim
N(0,\sigma^2)
$$

O termo:

$$
\varepsilon_i
$$

representa todos os fatores que afetam y e que não estão presentes no
modelo.

Se:

$$
E(\varepsilon_i)=0
$$

então:

$$
E(y_i)
=
\alpha+\beta x_i
$$

exatamente como desejado.

# 7. Interpretando o Termo de Erro

O termo de erro captura:

- fatores não observados;
- erros de medição;
- variabilidade individual;
- outros determinantes omitidos.

No exemplo vendas × preço, alguns fatores que poderiam estar
dentro do erro são:

- promoções;
- clima;
- concorrência;
- número de consumidores.

Portanto:

$$
\varepsilon_i
$$

representa tudo aquilo que influencia y mas não aparece explicitamente
no modelo.

# 8. Utilizando o Modelo para Previsão

Uma vez especificado o modelo:

$$
y_i
=
\alpha+\beta x_i+\varepsilon_i
$$

podemos realizar previsões.

A previsão ignora o erro específico da observação:

$$
\widehat y_i
=
\alpha+\beta x_i
$$

Nos slides foi proposta a pergunta:

> Qual seria a previsão de vendas para preço igual a 50?

> Qual seria a previsão para preço igual a 58?

Utilizando a reta ajustada visualmente:

$$
Price=50
\Rightarrow
Sales \approx 99
$$

e:

$$
Price=58
\Rightarrow
Sales \approx 85
$$

# 9. Mudanças Absolutas

Em um modelo linear:

$$
y
=
\alpha+\beta x
$$

a inclinação é:

$$
\frac{dy}{dx}
=
\beta
$$

Essa quantidade mede uma mudança absoluta.

Exemplo:

se:

$$
\beta=2
$$

então aumentar x em uma unidade aumenta y em aproximadamente duas
unidades.

# 10. Mudanças Relativas

Muitas decisões econômicas são tomadas em termos percentuais.

Suponha:

Loja A:

$$
150 \rightarrow 225
$$

Loja B:

$$
250 \rightarrow 400
$$

A mudança relativa é definida por:

$$
\frac{dx}{x}
$$

Para a Loja A:

$$
\frac{225-150}{150}
=
0.5
$$

ou:

$$
50\%
$$

Para a Loja B:

$$
\frac{400-250}{250}
=
0.6
$$

ou:

$$
60\%
$$

A Loja B apresentou o maior crescimento relativo.

# 11. Elasticidade

Em economia utilizamos frequentemente o conceito de elasticidade.

A elasticidade mede:

- a mudança percentual em y;
- causada por uma mudança percentual em x.

Formalmente:

$$
Elasticity
=
\frac{dy/y}{dx/x}
$$

Reorganizando:

$$
Elasticity
=
\frac{dy}{dx}
\frac{x}{y}
$$

Portanto:

$$
Elasticity
=
\beta\frac{x}{y}
$$

para um modelo linear simples.

# 12. Elasticidade no Modelo Linear

Considere:

$$
y
=
\alpha+\beta x
$$

Como:

$$
\frac{dy}{dx}
=
\beta
$$

a elasticidade é:

$$
Elasticity
=
\beta
\frac{x}{y}
$$

Observe que ela depende simultaneamente de:

- β;
- x;
- y.

Portanto a elasticidade não é constante ao longo da reta.

# 13. Motivação para Transformações Logarítmicas

Em muitas aplicações gostaríamos que a elasticidade fosse constante.

Isso torna a interpretação econômica muito mais simples.

Uma solução é utilizar transformações logarítmicas.

O modelo:

$$
\log(y)
=
\alpha
+
\beta\log(x)
+
\varepsilon
$$

possui uma propriedade extremamente conveniente:

a elasticidade é constante e igual a β.

# 14. Training Exercise 1.2

O objetivo deste treino é estudar a elasticidade
em diferentes especificações do modelo.

## Questão (a)

Mostrar que:

$$
\log(y_i)
=
\alpha
+
\beta\log(x_i)
+
\varepsilon_i
$$

possui elasticidade igual a β.

### Desenvolvimento

Partimos de:

$$
\log(y)
=
\alpha+\beta\log(x)
$$

Derivando:

$$
\frac1y
\frac{dy}{dx}
=
\beta
\frac1x
$$

Multiplicando ambos os lados por:

$$
x
$$

obtemos:

$$
\frac{dy}{dx}
\frac{x}{y}
=
\beta
$$

Pela definição:

$$
Elasticity
=
\frac{dy}{dx}
\frac{x}{y}
$$

Logo:

$$
\boxed{
Elasticity=\beta
}
$$

## Questão (b)

Determinar a elasticidade para:

$$
y_i
=
\alpha
+
\beta\log(x_i)
+
\varepsilon_i
$$

### Desenvolvimento

Derivando:

$$
\frac{dy}{dx}
=
\beta
\frac1x
$$

Portanto:

$$
Elasticity
=
\frac{dy}{dx}
\frac{x}{y}
$$

Substituindo:

$$
Elasticity
=
\left(
\beta\frac1x
\right)
\frac{x}{y}
$$

Logo:

$$
\boxed{
Elasticity
=
\frac{\beta}{y}
}
$$

A elasticidade depende do valor observado de y.

## Questão (c)

Determinar a elasticidade para:

$$
\log(y_i)
=
\alpha
+
\beta x_i
+
\varepsilon_i
$$

### Desenvolvimento

Derivando:

$$
\frac1y
\frac{dy}{dx}
=
\beta
$$

Portanto:

$$
\frac{dy}{dx}
=
\beta y
$$

Logo:

$$
Elasticity
=
\frac{dy}{dx}
\frac{x}{y}
$$

Substituindo:

$$
Elasticity
=
(\beta y)\frac{x}{y}
$$

Obtemos:

$$
\boxed{
Elasticity
=
\beta x
}
$$

A elasticidade varia com o nível de x.

# 15. Conclusões do Training

Obtivemos:

Modelo log-log:

$$
Elasticity=\beta
$$

---

Modelo linear-log:

$$
Elasticity=\frac{\beta}{y}
$$

---

Modelo log-linear:

$$
Elasticity=\beta x
$$

Esses resultados mostram como a interpretação econômica depende
da forma funcional adotada.
