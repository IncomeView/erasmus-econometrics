Nesta aula você aprenderá:

- como avaliar previsões obtidas por regressão;
- como construir intervalos de previsão;

- como construir testes e intervalos de confiança;

# 2. Previsão Pontual

Considere a regressão:

$$
y
=
a+bx
$$

Para um novo valor:

$$
x_0
$$

obtemos a previsão:

$$
\widehat y_0
=
a+bx_0
$$

Essa previsão recebe o nome de:

## previsão pontual

porque produz apenas um único valor.

# 3. O Valor Observado

O valor real é descrito pelo modelo:

$$
y_0
=
\alpha+\beta x_0+\varepsilon_0
$$

O problema é que:

$$
\varepsilon_0
$$

é desconhecido.

Assim,
mesmo que o modelo esteja correto,
o valor observado dificilmente coincidirá exatamente
com a previsão pontual.

# 4. Desvio Padrão Residual

Na aula anterior definimos:

$$
s^2
=
\frac{
\sum e_i^2
}{
n-2
}
$$

e:

$$
s
=
\sqrt{s^2}
$$

O valor:

$$
s
$$

mede a dispersão dos resíduos
ao redor da reta de regressão.

Quanto menor:

$$
s
$$

mais precisa tende a ser a previsão.

# 5. Intervalo de Previsão

Em vez de prever apenas um valor,
podemos prever uma faixa de valores plausíveis.

Assumimos:

$$
-ks
\le
\varepsilon_0
\le
ks
$$

Logo:

$$
\boxed{
\widehat y_0-ks
\le
y_0
\le
\widehat y_0+ks
}
$$

Esse é o intervalo de previsão apresentado na lecture.


# 6. Questão Conceitual da Aula

Considere dois intervalos:

$$
\widehat y_0 \pm s
$$

e:

$$
\widehat y_0 \pm 2s
$$

Qual deles possui maior probabilidade
de conter o valor observado?

### Discussão

Quanto maior:

$$
k
$$

maior o intervalo.

Portanto:

$$
\widehat y_0 \pm 2s
$$

possui maior probabilidade
de conter o valor observado.

Entretanto:

intervalos mais largos fornecem menos precisão.

Existe um compromisso entre:

- confiança;
- precisão.


# 7. As Sete Hipóteses da Regressão Simples

Todas as propriedades estatísticas da regressão
dependem de hipóteses sobre o processo gerador dos dados.

A lecture apresenta sete hipóteses.

## A1 — Processo Gerador dos Dados

O modelo verdadeiro é:

$$
y_i
=
\alpha+\beta x_i+\varepsilon_i
$$

A relação entre:

$$
y
$$

e:

$$
x
$$

é linear.


## A2 — Valores de x são Fixos

As observações:

$$
x_i
$$

são consideradas conhecidas e fixas.

Exemplo:

um gerente escolhe previamente
os preços que serão praticados.


## A3 — Erros com Média Zero

$$
E(\varepsilon_i)=0
$$

Não existe tendência sistemática
dos erros serem positivos ou negativos.


## A4 — Homocedasticidade

Todos os erros possuem a mesma variância:

$$
E(\varepsilon_i^2)
=
\sigma^2
$$

Cada observação carrega o mesmo nível de informação.


## A5 — Erros Não Correlacionados

$$
E(\varepsilon_i\varepsilon_j)
=
0
$$

para:

$$
i\neq j
$$

Os erros de diferentes observações
são independentes entre si.


## A6 — Parâmetros Constantes

Os parâmetros:

$$
\alpha
$$

e:

$$
\beta
$$

são fixos para todas as observações.

Existe apenas uma única reta populacional.


## A7 — Normalidade

Os erros possuem distribuição normal conjunta:

$$
\varepsilon_i
\sim
NID(0,\sigma^2)
$$

Essa hipótese será utilizada
para desenvolver testes estatísticos.


# 8. Objetivo da Avaliação Estatística

A partir das hipóteses A1–A7
queremos descobrir:

- média de b;
- variância de b;
- distribuição de b.

Em outras palavras:

queremos compreender as propriedades estatísticas
do estimador obtido por mínimos quadrados.

# 9. Relembrando o Estimador de Inclinação

Da aula anterior:

$$
b
=
\frac{
\sum (x_i-\bar x)(y_i-\bar y)
}{
\sum (x_i-\bar x)^2
}
$$

Essa expressão servirá de ponto de partida
para todas as demonstrações seguintes.


# 10. Preparação para a Próxima Parte

Na próxima seção:

- reescreveremos b em função dos erros;
- mostraremos que b é não viesado;
- derivaremos sua variância;
- construiremos a estatística t;
- obteremos intervalos de confiança.

Esses resultados formam a base teórica

# 11. Reescrevendo b em Função dos Erros

O objetivo central desta parte da aula é determinar
as propriedades estatísticas do estimador:

$$
b
$$

Para isso precisamos reescrevê-lo
diretamente em função dos erros:

$$
\varepsilon_i
$$

pois são eles que possuem propriedades estatísticas
definidas pelas hipóteses A1–A7.


# 12. Ponto de Partida

Partimos do estimador obtido na Lecture 1.3:

$$
b
=
\frac{
\sum (x_i-\bar x)(y_i-\bar y)
}{
\sum (x_i-\bar x)^2
}
$$

que será chamado de expressão (A).


# 13. Utilizando a Hipótese A1

Pela hipótese:

$$
y_i
=
\alpha+\beta x_i+\varepsilon_i
$$

temos:

$$
y_i-\bar y
=
(\alpha+\beta x_i+\varepsilon_i)
-
(\alpha+\beta\bar x+\bar\varepsilon)
$$

Cancelando:

$$
\alpha
$$

obtemos:

$$
y_i-\bar y
=
\beta(x_i-\bar x)
+
(\varepsilon_i-\bar\varepsilon)
$$

Esta será a expressão (B).


# 14. Substituindo (B) na Fórmula de b

Substituindo:

$$
y_i-\bar y
=
\beta(x_i-\bar x)
+
(\varepsilon_i-\bar\varepsilon)
$$

em:

$$
b
=
\frac{
\sum (x_i-\bar x)(y_i-\bar y)
}{
\sum (x_i-\bar x)^2
}
$$

obtemos:

$$
b
=
\beta
+
\frac{
\sum (x_i-\bar x)
(\varepsilon_i-\bar\varepsilon)
}{
\sum (x_i-\bar x)^2
}
$$


# 15. Eliminando a Média dos Erros

Observe:

$$
\sum(x_i-\bar x)\bar\varepsilon
=
\bar\varepsilon
\sum(x_i-\bar x)
$$

mas:

$$
\sum(x_i-\bar x)=0
$$

Logo:

$$
\bar\varepsilon
\sum(x_i-\bar x)
=
0
$$

Assim:

$$
b
=
\beta
+
\frac{
\sum(x_i-\bar x)\varepsilon_i
}{
\sum(x_i-\bar x)^2
}
$$


# 16. Definição dos Pesos cᵢ

Definimos:

$$
c_i
=
\frac{
x_i-\bar x
}{
\sum(x_i-\bar x)^2
}
$$

Logo:

$$
\boxed{
b
=
\beta
+
\sum c_i\varepsilon_i
}
$$

A partir deste ponto todas as propriedades
de:

$$
b
$$

podem ser obtidas a partir das propriedades dos erros.


# 17. Interpretando a Expressão Obtida

O resultado:

$$
b
=
\beta+\sum c_i\varepsilon_i
$$

mostra que:

- b não é igual exatamente a β;
- b oscila ao redor de β;
- essas oscilações dependem dos erros da amostra.

A questão fundamental passa a ser:

qual é o valor médio dessas oscilações?

# 18. Não Viesamento de b

Tomando esperança:

$$
E(b)
=
E
\left(
\beta+\sum c_i\varepsilon_i
\right)
$$

Pela linearidade:

$$
E(b)
=
E(\beta)
+
\sum E(c_i\varepsilon_i)
$$


# 19. Aplicando as Hipóteses

Pela hipótese A6:

$$
\beta
$$

é constante.

Portanto:

$$
E(\beta)=\beta
$$

Pela hipótese A3:

$$
E(\varepsilon_i)=0
$$

Como:

$$
c_i
$$

é constante:

$$
E(c_i\varepsilon_i)
=
c_iE(\varepsilon_i)
=
0
$$


# 20. Resultado Fundamental

Obtemos:

$$
E(b)
=
\beta
$$

ou:

$$
\boxed{
E(b)=\beta
}
$$

Conclusão:

## b é um estimador não viesado de β.

Em média,
o método dos mínimos quadrados acerta
o verdadeiro valor do parâmetro.


# 21. Indo Além da Média

Saber que:

$$
E(b)=\beta
$$

é importante.

Entretanto:

dois estimadores podem possuir a mesma média
e apresentar níveis muito diferentes de incerteza.

Precisamos então calcular:

$$
Var(b)
$$

# 22. Variância de b

Partimos novamente de:

$$
b
=
\beta
+
\sum c_i\varepsilon_i
$$

Como:

$$
E(b)=\beta
$$

temos:

$$
b-E(b)
=
\sum c_i\varepsilon_i
$$

Portanto:

$$
Var(b)
=
E
\left[
\left(
\sum c_i\varepsilon_i
\right)^2
\right]
$$


# 23. Expandindo a Variância

Expandindo:

$$
Var(b)
=
E
\left[
\sum_i
\sum_j
c_ic_j
\varepsilon_i
\varepsilon_j
\right]
$$

Como:

$$
c_i
$$

são constantes:

$$
Var(b)
=
\sum_i
\sum_j
c_ic_j
E(\varepsilon_i\varepsilon_j)
$$


# 24. Aplicando a Hipótese A5

Pela hipótese:

$$
E(\varepsilon_i\varepsilon_j)=0
$$

para:

$$
i\neq j
$$

todos os termos cruzados desaparecem.

Restam apenas:

$$
E(\varepsilon_i^2)
$$



# 25. Aplicando a Hipótese A4

Pela homocedasticidade:

$$
E(\varepsilon_i^2)
=
\sigma^2
$$

Logo:

$$
Var(b)
=
\sigma^2
\sum c_i^2
$$


# 26. Resultado Final

Após substituir:

$$
c_i
=
\frac{x_i-\bar x}
{\sum(x_i-\bar x)^2}
$$

obtemos:

$$
\boxed{
\sigma_b^2
=
\frac{
\sigma^2
}{
\sum(x_i-\bar x)^2
}
}
$$

Este é um dos resultados mais importantes
da teoria da regressão simples.


# 27. Interpretação da Fórmula

Observe:

$$
\sigma_b^2
=
\frac{
\sigma^2
}{
\sum(x_i-\bar x)^2
}
$$

Quanto maior a variabilidade dos valores de x:

$$
\sum(x_i-\bar x)^2
$$

menor será a variância de:

$$
b
$$

e mais precisa será a estimativa.

Este resultado tem enorme importância prática.

Ele mostra que amostras com pouca variação em x
produzem estimativas menos precisas.

# 30. Aplicando a Hipótese A7

\varepsilon_i

$$

\sum c_i\varepsilon_i

Obtemos:

\boxed{

\sim

\left(
\beta,

# 32. Padronizando b

De forma semelhante ao que fazemos com qualquer

podemos padronizar:

$$

$$

usando sua média e desvio padrão.

Definimos:

$$
Z
=
\frac{

}{
\sigma_b
}

O resultado é:

$$
\boxed{
Z
\sim
N(0,1)
}
$$

Isto significa que:

- média zero;
- variância igual a um.

A estatística Z permite avaliar

$$

$$
\beta
$$

A estatística:

contém:

$$

que depende de:

$$

$$

é desconhecida.

# Questão (f)

Utilizar:

para estudar o viés quando:

- SN = 3

assumindo:

```python
import pandas as pd

beta = 1

sn_values = [1, 3, 10]

results = []

for sn in sn_values:

    bias = -(beta)/(sn + 1)

    results.append(
        {
            "Signal Noise Ratio": sn,
            "Approximate Bias": bias
        }
    )

pd.DataFrame(results)
```
